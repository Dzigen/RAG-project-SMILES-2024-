```python
import os
import pandas as pd

from utils import parse_gen
```

```python
logs_dir = "logs_path"
prefix = "answer_labeling_lmjudged_raw"
endings = ["Modellbewertung", "Punktzahl"]
# endings = ["मॉडल स्कोर", "अंक"]
# endings = ["Điểm số mô hình", "Điểm"]
# endings = ["Score", "Model score"]
# endings = ["Оценка модели", "Оценка"]
save_prefix = "answer_labeling_lmjudged"
raw_score_colname = "model_assessment"
```

```python
for filename in sorted(os.listdir(logs_dir)):
    if filename.startswith(prefix):
        exp_num = os.path.splitext(filename)[0][-7:]
        if exp_num[0].isdigit():
            
            filepath = os.path.join(logs_dir, filename)
            logs = pd.read_csv(filepath)
            logs["parsed_responses"] = logs[raw_score_colname].apply(lambda x: parse_gen(x, endings=endings))
            save_path = os.path.join(logs_dir, save_prefix + "_" + exp_num + ".csv")
            logs.to_csv(save_path, index=False)
            print(exp_num)
            print((logs["parsed_responses"] > 0).mean())
```
