```python
import os
import pandas as pd
```

```python
labeling_dir = "labeling/mtssquad"
exp_dir = "mtssquad/results"
exp_nums = [
    "1_1", 
    "1_2",
    "1_3",
    "2_1",
    "2_2",
    "2_3",
    "4_1",
    "4_2",
    "4_3"
]
label_data = {}
for filename in os.listdir(labeling_dir):
    for exp_num in exp_nums:
        if exp_num in filename:
            filepath = os.path.join(labeling_dir, filename)
            data = pd.read_csv(filepath)
            label_data[exp_num] = data.iloc[:150]

exp_data = {}
for filename in os.listdir(exp_dir):
    for exp_num in exp_nums:
        if "answer_labeling_" + exp_num + "_1_1" in filename:
            filepath = os.path.join(exp_dir, filename)
            data = pd.read_csv(filepath)
            exp_data[exp_num] = data
```

```python
new_labels = {}
save_prefix = "мтс разметка ответов - data "
for exp_num in exp_nums:
    label = label_data[exp_num]
    exp = exp_data[exp_num]
    matching_set = label.merge(exp, on="Question", how="left")
    label["llm_responses"] = matching_set["model_assessment"]
    save_path = os.path.join(labeling_dir, save_prefix + exp_num + ".csv")
    label.to_csv(save_path)
```
