```python
import os
import pandas as pd
```

```python
logs_dir = "mtssquad/results"
prefix = "MTS_"
```

```python
for filename in os.listdir(logs_dir):
    old_file_path = os.path.join(logs_dir, filename)
    new_file_path = os.path.join(logs_dir, filename[len(prefix):])
    os.rename(old_file_path, new_file_path)
```

```python
for filename in os.listdir(logs_dir):
    old_file_path = os.path.join(logs_dir, filename)
    name, ext = os.path.splitext(filename)
    new_file_path = os.path.join(logs_dir, name + "_1_1" + ext)
    os.rename(old_file_path, new_file_path)
```
