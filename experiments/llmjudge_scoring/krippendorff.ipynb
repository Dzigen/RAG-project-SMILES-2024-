```python
import os
import pandas as pd
import numpy as np
from scipy.stats import pearsonr
import krippendorff as kp
```

```python
dataset_name = "squad2"
data_dir = "/home/jovyan/work/alexander_workspace/exp/labeling/"
dataset_dir = os.path.join(data_dir, dataset_name)
res = {}
for exp in sorted(filter(lambda x: "ipynb" not in x, os.listdir(dataset_dir)))[:9]:
    dataset_path = os.path.join(dataset_dir, exp)
    data = pd.read_csv(dataset_path)
    cols = list(filter(lambda x: "_assessment" in x, data.columns))
    inds = np.arange(150)
    labels = data[cols].iloc[inds]
    majority = labels.mode(axis=1).squeeze().values.astype(int)
    llm_responses = data["llm_responses"].iloc[inds].fillna(0).values.astype(int)
    pearson = pearsonr(llm_responses, majority).statistic
    kp_alpha = kp.alpha(reliability_data=labels.values.T.astype(int))
    res[exp] = {"kp_alpha": kp_alpha, "pearson_corr": pearson}
for k, v in res.items():
    new_v = {metric_name: round(metric_val, 3) for metric_name, metric_val in v.items()}
    print(f"{k}: {new_v}")
print("mean kp_alpha:", round(np.mean([v["kp_alpha"] for v in res.values()]), 3))
print("mean pearson_corr:", round(np.mean([v["pearson_corr"] for v in res.values()]), 3))
```
