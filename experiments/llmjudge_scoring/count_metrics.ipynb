```python
import os
import pandas as pd
import numpy as np
```

```python
path_to_res_dir = "/home/jovyan/work/alexander_workspace/exp/mtsquad/results"
criterion = lambda x: "lmjudged" in x
score_field = "parsed_responses"
metric_fn = np.mean
nrows = 150
```

```python
dfs = dict()
for filename in os.listdir(path_to_res_dir):
    if criterion(filename):
        path_to_df = os.path.join(path_to_res_dir, filename)
        df = pd.read_csv(path_to_df, nrows=nrows)
        exp_num = "_".join(os.path.splitext(filename)[0].split("_")[-2:])
        dfs[exp_num] = metric_fn(df[score_field])
```

```python
for k, v in sorted(dfs.items()):
    print(f"{k}: {round(v, 3)}")
```

```python
report = pd.DataFrame(columns=["exp_num", "mean_score"])
for k, v in sorted(dfs.items()):
    report.loc[len(report)] = ({"exp_num": k, "mean_score": round(v, 3)})
report
```

```python
report.to_csv("report_lmjudge.csv", index=False)
```
