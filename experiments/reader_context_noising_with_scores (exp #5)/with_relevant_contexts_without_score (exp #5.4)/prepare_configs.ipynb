```python
from copy import copy
import json
```

```python
PARAMS = {
    'version': ...,
    'num_samples': 2000,
    'num_contexts': ...,
    'model': ...,
    'bp': ...,
    'scores': {'rel': 1.0, 'unrel': 0.0},
    'gen_strat': {'max_new_tokens': 1024, 'do_sample': False, 'num_beams': 1},
    'calculate_entropy': True,
    'revert': ...,
    'centered': ...,
    'LOGS_SAVE_DIR': ...,
    'QA_DATASET_PATH': ...,
    'CONTEXTS_DATASET_PATH': ...
}
```

```python
SRC_DIR = '/home/jovyan/work/repo_name'
BASE_DIR = '/home/jovyan/work/repo_name/experiments/reader_context_noising_with_scores (exp #5)/with_relevant_contexts_without_score (exp #5.4)'

AVAILABLE_MODELS = [f"{SRC_DIR}/models/Undi95/Meta-Llama-3-8B-Instruct-hf", f"{SRC_DIR}/models/Qwen/Qwen2.5-7B-Instruct"]
AVAILABLE_LANGUAGES = {
    'de': {
        'qa': f"{SRC_DIR}/data/mlqa/de/qa_dataset.csv",
        'contexts': f"{SRC_DIR}/data/mlqa/de/contexts.csv",
        'logs': f'{BASE_DIR}/mlqa/de/logs'
    },
    'hi': {
        'qa': f"{SRC_DIR}/data/mlqa/hi/qa_dataset.csv",
        'contexts': f"{SRC_DIR}/data/mlqa/hi/contexts.csv",
        'logs': f'{BASE_DIR}/mlqa/hi/logs'
    },
    'vi': {
        'qa': f"{SRC_DIR}/data/mlqa/vi/qa_dataset.csv",
        'contexts': f"{SRC_DIR}/data/mlqa/vi/contexts.csv",
        'logs': f'{BASE_DIR}/mlqa/vi/logs'
    },
    # 'zh': {
    #     'qa': f"{SRC_DIR}/data/mlqa/zh/qa_dataset.csv",
    #     'contexts': f"{SRC_DIR}/data/mlqa/zh/contexts.csv",
    #     'logs': f'{BASE_DIR}/mlqa/zh/logs'
    # },
    # 'ar': {
    #     'qa': f"{SRC_DIR}/data/mlqa/ar/qa_dataset.csv",
    #     'contexts': f"{SRC_DIR}/data/mlqa/ar/contexts.csv",
    #     'logs': f'{BASE_DIR}/mlqa/ar/logs'
    # }
}
BASE_CONFIGS = [['1.1', 5, False, False], ['2.1', 5, True, False], ['3.1', 5, False, True]]
```

```python
CONFIGS_DIR = './configs'
configs_counter = 1

for lang, paths in AVAILABLE_LANGUAGES.items():
    for b_config in BASE_CONFIGS:
        for i, model_path in enumerate(AVAILABLE_MODELS):
            tmp_params = copy(PARAMS)
            tmp_params['version'] = f"{b_config[0]}.{i+1}"

            tmp_params['bp'] = lang
            tmp_params['LOGS_SAVE_DIR'] = paths['logs']
            tmp_params['QA_DATASET_PATH'] = paths['qa']
            tmp_params['CONTEXTS_DATASET_PATH'] = paths['contexts']

            tmp_params['model'] = model_path
            tmp_params['revert'] = b_config[2]
            tmp_params['centered'] = b_config[3]
            tmp_params['num_contexts'] = b_config[1]
            
            with open(f"{CONFIGS_DIR}/config{configs_counter}_{lang}.json", 'w', encoding='utf-8') as fp:
                fp.write(json.dumps(tmp_params, ensure_ascii=False, indent=1))

            configs_counter += 1
```
