```python
from collections import Counter
import matplotlib.pyplot as plt
from datasets import load_dataset
import numpy as np
import pandas as pd

CONTEXTS_DATASET_PATH = "../../data/squadv2/contexts.csv"
```

```python
contexts_df = pd.read_csv(CONTEXTS_DATASET_PATH)
```

```python
cntx_lens = list(map(lambda cntx: len(cntx), list(set(contexts_df['context'].tolist()))))
print("unique cntxs: ", len(cntx_lens))
print(f"median/std: {np.median(cntx_lens)}/{np.std(cntx_lens)}")
plt.hist(cntx_lens)
plt.show()
```
